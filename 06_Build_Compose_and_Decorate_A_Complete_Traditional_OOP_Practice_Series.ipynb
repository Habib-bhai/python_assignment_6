```python
# 1. Using self

class student:
  def __init__(self, name, marks) -> None:
    self.name = name
    self.marks = marks

  def display(self):
    print(f"Student '{self.name}' has scored: {self.marks}")

student1 = student("John", 90)
student1.display()

```

```python
# 2. Using cls

class counter:
  count = 0

  def __init__(self, name):
    self.name = name
    counter.count +=1

  @classmethod
  def get_instance_count(cls):
    print(cls)
    # accessing it using "cls", not hardcoding it here.
    return f"Count of instances: {cls.count}"


obj1 = counter("obj1")
obj2 = counter("obj2")
print(counter.get_instance_count())
```

```python
# 7. Access Modifiers: Public, Private, and Protected
class Employee:
  def __init__(self, name, salary, ssn):
      self.name = name
      self._salary = salary
      self.__ssn = ssn


employee_object = Employee("habib", 80000, 123)
print(employee_object.name) # public attribute, no error, occured.
print(employee_object._salary) # protected attribute, no error occured, because python allows to access attributes with "_" protected identifier, we have to take care of this convention on our own.
print(employee_object.__ssn) # gives error, because python changes the name of the private attribute through "Name Mangling"
```

```python
# 8. The super() Function

class Person:
  def __init__(self, name):
    self.name = name


class Teacher(Person):
  def __init__(self, name, subject):
    super().__init__(name)
    self.subject = subject

teacher_object = Teacher("habib", "computer science")

print(teacher_object.name)
print(teacher_object.subject)
```

```python
# 9. Abstract Classes and Methods
from abc import ABC, abstractmethod

class Shape(ABC):
  @abstractmethod
  def area(self,l, b):
    pass


class Rectangle(Shape):
  def area(self, l , b):
    print("Area of Rectangle: ", l * b)


rect_obj = Rectangle()
rect_obj.area(10, 20)
```

```python
# 13. Composition
# composition refers to "part of" relationship.
# Where one class owns some else class i.e. another class,
# gets initialized in the class, so the life time of that class
# is tied to the lifetime of the container class.


class Engine:
  def __init__(self, engine_type, max_speed):
    self.engine_type = engine_type
    self.max_speed = max_speed

  def start(self):
    print(f"brummmm brummm, {self.engine_type} can go up to {self.max_speed}")


class car:
  def __init__(self, model ):
    self.engine = Engine("v8", 400)
    self.model = model

  def start(self):
    print(f"Car model: {self.model}")
    self.engine.start()



car_obj = car("bugatti chiron")

car_obj.start()
```

```python
# 14. Aggregation
# This refers to a "has a" relationship, where one class just has a relationship
# with the other which doesn't effect anyone's independant existance.


class Professor:
  def __init__(self, name, subject):
    self.name = name
    self.subject = subject

  def professor_teaches(self):
    print(f"{self.name} teaches {self.subject}")

class University:
  def __init__(self, name):
    self.name = name
    self.professor = []

  def add_professors(self, professor):
    self.professor.append(professor)

  def display_professors(self):
    print(f"University: {self.name}")
    for professor in self.professor:
      print(professor.name)


# we can observe here professor and university have relation
# but professors can exist independantly as well.

professor_1 = Professor("Habib", "Maths")
professor_2 = Professor("Anees", "Reproduction")

uni = University("University of Mads")

uni.add_professors(professor_1)
uni.add_professors(professor_2)

uni.display_professors()
```

```python
# 15. Method Resolution Order (MRO) and Diamond Inheritance

class A:
  def show(self):
    print("Hello from A")

class B:
  def show(self):
    print("Hello from B")

class C(A):
  def show(self):
    print("Hello from C")


class D(B, C):
  pass

obj = D()

print(D.mro())
obj.show()
```

```python
# 16. Function Decorators

def log_function_call(func):
  def wrapper():
    print(f"{func.__name__}  Function is being called ")
    func()
  return wrapper


@log_function_call
def say_hello():
  print("hello world")

say_hello()
```

```python
# 17. Class Decorators


def add_greeting(cls):
    cls.greet = lambda self: "Hello from Decorator!"
    return cls

@add_greeting
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def details_displayer(self):
      print(self.name, self.age)

p1 = Person("Habib", 19)

p1.greet()
```

```python
# 18. Property Decorators: @property, @setter, and @deleter

class Product:
  def __init__(self, price):
    self._price = price

  # method that work as a property, i.e. just is responsible to show a value
  @property
  def price(self):
    return self._price

  # method, but working as a variable that can be reinitialized.
  @price.setter
  def price(self, price):
    self._price = price
    print(self._price)

  @price.deleter
  def price(self):
    print("DELETE HOGAYA OYEEEEEEEEEE")



product_1 = Product(5500)

print(product_1.price)

product_1.price = 2000

del product_1.price

```

```python
# 21. Make a Custom Class Iterable

class Countdown:
  def __init__(self, starter_number):
    self.starter_number = starter_number

  def __iter__(self):
    return self

  def __next__(self):
    value = self.starter_number
    if value < 0:
      raise StopIteration
    self.starter_number -= 1
    return value


counter_obj = Countdown(10)

obj_iterable = iter(counter_obj)

print(obj_iterable.__next__())
print(obj_iterable.__next__())
print(obj_iterable.__next__())
print(obj_iterable.__next__())
print(obj_iterable.__next__())
print(obj_iterable.__next__())
print(obj_iterable.__next__())
print(obj_iterable.__next__())
print(obj_iterable.__next__())
print(obj_iterable.__next__())
print(obj_iterable.__next__())
```
